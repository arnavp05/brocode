```python
# Import required libraries
!pip install gensim
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from gensim.models import Word2Vec
import pandas as pd
```

```python
# Sample sentences
documents = [
    "I love natural language processing",
    "Natural language processing is interesting",
    "I love machine learning"
]
```

```python
# 1. Bag of Words - Count Occurrence
# ---------------------------------------------------

print("1. Bag of Words - Count Occurrence\n")

# Create CountVectorizer object
vectorizer = CountVectorizer()

# Convert text into count matrix
X = vectorizer.fit_transform(documents)

# Convert into table format
bow_table = pd.DataFrame(X.toarray(), columns=vectorizer.get_feature_names_out())

print(bow_table)
```

```python
# 2. Normalized Count Occurrence
# ---------------------------------------------------

print("\n2. Normalized Count Occurrence\n")

# Normalize by dividing each count by total words in sentence
normalized_bow = bow_table.div(bow_table.sum(axis=1), axis=0)

print(normalized_bow)
```

```python
# ---------------------------------------------------
# 3. TF-IDF
# ---------------------------------------------------

print("\n3. TF-IDF\n")

# Create TF-IDF Vectorizer
tfidf = TfidfVectorizer()

# Transform documents
tfidf_matrix = tfidf.fit_transform(documents)

# Convert into table
tfidf_table = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

print(tfidf_table)
```

```python
# 4. Word2Vec Embeddings
# ---------------------------------------------------

print("\n4. Word2Vec Embeddings\n")

# Tokenize sentences
tokenized_sentences = [sentence.split() for sentence in documents]

# Train Word2Vec model
model = Word2Vec(
    tokenized_sentences,
    vector_size=5,   # Size of word vector
    window=2,
    min_count=1
)
```

```python
# Display vector for word "love"
print("Vector for word 'love':\n")
print(model.wv['love'])
```
